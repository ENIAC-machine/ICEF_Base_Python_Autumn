# Problem 4 (20 pts). Monte-Carlo of e

During the class/ office hours we estimated $\large \pi$ via the Monte-Carlo method, now it is your objective to estimate the exponent ($\large e\approx 2.71828182846$)\
It just so happens that E[ξ]=e, where ξ is a random variable that is defined as follows:\
$\large \xi = \underset{n}{\text{argmin}} ( \sum_{i=1}^n r_i > 1 ), r_i \textasciitilde  U[0, 1]$\
Constraints:
1. $\large N \in (0, 2^{32})$

````python
def find_e(N: int= 1_000_000)-> float | None:
    '''
    
    Computes the Monte-Carlo estimate of the exponent

    Inputs:
        N: int - number of samples

    Outputs:
        e_approx: float| None - the estimate of the exponent
        
    '''
    ###your code for assertion

    #initialize the array of zeros to house all the samples
    vals = np.zeros(N)### your code here

    for i in range(N):
        #initialize the number of generated values and their sum 
        ### your code here
        n, sm = 0, 0
        #begin the while loop for each sample of xi
        while sm <= 1:
            #add the uniform random variable
            sm += np.random.uniform()
            n += 1
            #increment n
            ###your code here
        
        vals[i] = n

    #estimate the expectation of the variable via mean
    e_approx = vals.mean()### your code here
    
    return e_approx
````
