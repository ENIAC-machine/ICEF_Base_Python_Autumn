## Problem 1

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Constraints:
1. 1 $\large \le$ n $\large \le$ 45

````python
def factorial(n: int) -> int:
    '''
    
    Compute n!

    Inputs:
        n: int - some positive number n

    Outputs:
        res: int - the factorial of n
    
    '''
    assert isinstance(n, int) and n >= 0, "Wrong input n"
    
    res = 1
    
    for i in range(2, n+1):
        res *= i
    
    return res

def num_ways_to_reach_top(n: int)-> int:
    '''
    Computes the number of ways to reach the top of the staircase

    Inputs:
        n: int - number of stairs in the staircase

    Outputs:
        n_ways: int - number of ways to reach the top given that we can step over 1 or 2 steps

    '''
    assert isinstance(n, int) and 1 <= n <= 45, f"n is of wrong type or out of bounds, expected int from 1 to 45, got {type(n), n}"

    n_ways = 1 #initially we have the first unique way of going only 1 step at a time
    way = '1'*n #the way in question

    #the idea here is that the way is represented as string and we have to calculate the number of combinations of 2's among 1's
    #for each possible number of 2's in the string

    #begin iterations, think about how many iterations are going to be there
    for i in range(n // 2):
        way = way[2:] + '2' ###your code here to add the 2 to the end of the way and subtract the first 1's from the beginning in one line 
        n = len(way) #total number of elements
        k = way.count("2")###your code to count 2's in the string
        n_ways += factorial(n) / (factorial(k) * factorial(n-k))###your code to implement the formula of finding Combinations of k among n elements 
    
    return int(n_ways)
````

````python
#code for self check
for n, res in tqdm(zip([2,3,4,5,6,7,10], [2,3,5,8,13,21,89])):
    assert num_ways_to_reach_top(n) == res, "Test not passed"
````

## Problem 2

In this task you should implement a Monte-Carlo algorithm to find the root of a number ```n```

Constraints:
1. n $\large \ge$ 0

*Hint: what happens when you square all values sampled from U[0, 1]?*

````python
root = lambda n, N=1_000_000 : np.sum(np.random.uniform(0, n, N) ** 2 <= n) / N * n
````

````python
#code for self check
for _ in tqdm(range(1_000)):
    check_num = np.random.randint(0, 100)
    assert np.allclose(check_num**.5, root(check_num), atol=3)
````

## Problem 3

Given a 32-bit integer ```n``` you need to reverse its bit order and convert it back to corresponding integer, see examples below:\
4 -> 00000000000000000000000000000100\
2**29 -> 00100000000000000000000000000000

43261596 ->	00000010100101000001111010011100\
964176192 ->	00111001011110000010100101000000

*Hint: remember the properties of string, how to manipulate with it (concatinate, etc.)*

````python
def bit_reverse(n: int) -> int:
    '''
    Reverses the bits of the positive 32-bit integer returns the corresponding integer result

    Inputs:
        n: int - 32-bit integer

    Outputs:
        n_rev: int - integer representation of reversed bits

    '''
    assert isinstance(n, int), "Wrong input for n"

    bin_repr = bin(n)[2:] #get the bitwise representation of n
    bin_repr = ('0'*(32 - len(bin_repr)) + bin_repr)[::-1]
    n_rev = int(bin_repr, 2) #get the integer representation of the bit string
    return n_rev
````

````python
#code for you to test yourself
for inpt, res in zip([43261596, 2147483644, 2, 1 << 30], [964176192, 1073741822, 1 << 30, 2]):
    assert bit_reverse(inpt) == res, "You got a mistake!"
````

## Problem 4

Write an algorithm to determine if a number ```n``` is happy.

A happy number is a number defined by the following process:

Starting with any positive integer, replace the number by the sum of the squares of its digits\
Repeat the process until the number equals 1 (where it will stay), or it loops endlessly in a cycle which does not include 1\
Those numbers for which this process ends in 1 are happy\
Return true if n is a happy number, and false if not

Constraints:
1. n > 0

````python
def is_happy(n: int) -> bool:
    '''
    
    Checks if the number is happy or not
    
    '''
    assert isinstance(n, int) and n > 0, "Wrong input"

    #define the array to hold the past results
    vals = {n}
    
    #define the variable to hold the current value while applying transformations, starting with n
    num = n

    while num != 1:

        #get the digits 
        digits = list(str(num))

        #perform the transformation
        num = sum([int(digit)**2 for digit in digits])

        #break out of the loop if we are going in cycles
        if num in vals:
            break

        
        vals.add(num)
    
    return True if 1 in vals else False
````

````python
#code for self check
for inpt, res in tqdm(zip([1, 2, 19, 20, 10_001, 42069, 87654], [True, False, True, False, False, False, True])):
    assert is_happy(inpt) == res, "You got a mistake!"
````
