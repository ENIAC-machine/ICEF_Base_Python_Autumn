```python
def filter_(char_list:list, filename:str)->list:
    '''
    filter out the unneccesary values from the file

    Inputs:
        char_list: list - list of characters to filter out from the file
        filename: str - the name of the file to open
    
    Output:
        out:list - list of filtered lines of the file
    '''
    assert type(char_list) == list, "Wrong type of variable" #optional

    out = None

    try:

        with open(f'{filename}.txt', "r") as file:
            lines = file.readlines() #read the lines 
    
        out = [] #redefine output as a list
        for idl, line in enumerate(lines): #iterate over lines in the file
            lines[idl] = list(line) #convert each line into list
        for idl, line in enumerate(lines): #iterate over lines in the file
            for idc, character in enumerate(line): #iterate over characters in line
                if character in char_list: #check if the character is in the char_list 
                    lines[idl][idc] = "" #change the character to empty
                    print("changed the value!")
            out.append(line) #add it to the output
        
    except:
        print("Some error hapenned during the computation!")

    finally:
        return out
```

```python
def derivative(y:list, x:list=None)->list:
    '''
    Inputs:
        y:list - list of values of the function
        x:list - list of argument values of the function

    Outputs:
        der:list - list of derivatives

    '''

    if x == None:
        x = list(range(len(y)))  
        
    if type(y) != list or type(x) != list:
        try:
            y = list(y)
            x = list(x)
        except:
            print("x and/or y of wrong type!")
            return None
    
    der = [0]
    types = {float, int}
    
    for idx in range(1, len(x)):
        if type(y[idx]) in types and type(x[idx]) in types and type(y[idx-1]) in types and type(x[idx-1]) in types: 
            derivative_val = (y[idx] - y[idx-1]) / (x[idx] - x[idx-1])
            der.append(derivative_val)
        else:
            print(f"Wrong values as inputs, expected types of float/int, got types of {type(y[idx]), type(x[idx]), type(y[idx-1]), type(x[idx-1])}")
            return None
    return der
```

```python
def writer(filename:str, lines:list)->None:
    '''
    Inputs:
        filename:str - the name of the file we write in
        lines:list the lines we write

    Outputs:
        None
    '''

    for idx, line in enumerate(lines):
        lines[idx] = str(line)

    try:
        with open(filename, "r+") as file:
            file.writelines(lines)
    except:
        print("Error! File doesn't exist")
    else:
        print("Writing went well!")
```
