### Concept of (im-)mutability

The concept of (im-)mutability is very simple- it just refers to the fact whether a variable can or can **not** be changed. Mutable means changable,\
immutable- the opposite. This may seem unimportant because why would it be? However, it plays a much bigger role than you think, so I'm going to\
give you a couple of tips regarding working with (im-)mutable variables. Before that though, keep in mind that most of the variables in python are **immutable** (```int```, ```str```, ```float```, ```tuple```, etc.). The key mutable types for you right now are ```list```, ```dict``` and ```set``` (well also ```bytearray``` but that's not important right now). One of the worst mistakes one could do with mutable datatypes is to make them a default\
value in a function. For example, we have this function:
```python
def example_func(arr_arg=[]):
    arr_arg.append(1) #this function appends the value to the end of the array
    return arr_arg

```
Because you passed a mutable datatype to the function default argument that's created with it and isn't discarded until the end of script execution\
every time you call this function, the ```arr_arg``` list will grow and not be just ```[1]``` as intended in the code. Of course this particular case\
is just a demonstration, but the problem is very real, better to know ahead than to find out during debugging\

## Loops

Why? А главное- зачем?\
If you'll continue to work in the field of programming, you will ***always*** need to loop over something or at least there will be an implication that\
a loop is required here. Examples? Iterate over data to update it, put variables into an array, check that the value you need is in the array etc.\
So how do loops work in python? It's easy, firstly, there are two types of loops: the ```while``` loop and the ```for``` loop. Difference? The while loop checks some user-defined condition at every step, while the for loop just iterates over some array you give it.\
It's ***heavily*** recommended (borderline mandatory) for you to use the ```for``` loops because they are *much* faster. You'd be surprised how long\
it may take to check a condition, especially how much time it eats up during the entire run of the loop. For the educational purposes, we will practice both\
Here's the syntax for them with examples:

````python
#let's define some sample array
arr = ['first element', 'second element', 'yolo']
````

````python
#syntax of the for loop
for element in arr: #here you 'take' each element from array "arr" and below define the operation to perform on it
    print(element) #in this case it will just print each element from the array, btw print is the function that outputs what you give it onto your screen
````

````python
#you can also break the loop with the 'break' keyword:

for element in arr:
    print(element)
    break #in this case we'll print only the first element of the array, as the loop will break right after that
````

````python
#if you want/need to see the ids of the elements you iterate over, use python's native keyword 'enumerate'
for idx, element in enumerate(arr): #here idx is the element's id in arr
    print(idx, element)
````

````python
#while loop syntax, in this case only the first 2 elements will be printed
iterator = 0 #because this is a while loop we have to manually define the iterator and iterate over the array
while element != 'second element': #in this case the condition is that variable 'element' should NOT have the value 'yolo'
    element = arr[iterator] #asign the variable element with the value from array
    iterator += 1 #add 1 to iterator, this is equivalent to iterator = iterator + 1
    print(element, iterator)
````

## Try block

There will be times (very very many times), when the scripts you write will throw an error, however in a small percentage of these times, it may be expected.\
For such cases python has the ```try``` block. What does it do? Error handling basically, it redirects the flow of your script in case the code inside\
threw an error. You're right to consider that it's quite similar to the previous ```if``` statement, (it's even compatible with the ```else``` keyword!).\
So what parts does this block consist of?\
1. ```try``` - in this block the first instance of the script is executed, if it doesn't throw an error, execution continues to the ```else``` and ```finally``` blocks
2. ```except``` - used for error handling, can be used as a function with errors as input. Executes if ```try``` block threw an error that's one of it's arguments. If no arguments are passed, executes after any error
3. ```else```  - follows ```except```. Executes after the ```try``` block in case there was no error, used for finalizing programm in case no errors arose
4. ```finally``` - always executes after all of the blocks above. Used for actions that must be done independent of the script flow

I know, it may be a lot to take in at once, so I gave you example usage and syntax below to (hopefully) make it more understandable for you:
```python

num1 = 15
num2 = int(input('Enter some number to divide 15 by: ')) #we convert it to int as by default any input is str
num3 = 0

try:
    num3 = num1 / num2 #try to divide num1 by num2 and put that value in num3
except:
    print('Well, something went wrong') #in case something goes wrong (like division by zero) print out this phrase
else:
    print(num3) #in case nothing goes wrong print the result of the division
finally:
    num3 = 0 #independent of the division result reset the num3 variable to 0

```

## Functions and decorators

Lets again start with the most fundamental stuff: when exactly do you need to make a function?\
Well, *strictly speaking* one should make the function when the operation(-s)/algorithm(-s) in it are expected to be used(called)\
However, at times (especially when writing large scripts) it's more convenient to write every step of the process/programm as a function\
in order to manage memory and get rid of unneccesary variables at each step of the process.\
Let's now see how to work with functions in python:\
Firstly, we define the function via the ``` def ``` keyword, like ```def example_function():``` would define a function with the name "example_function".\
You define the arguments the function accepts in the braces, like ```def example_function(arg1):``` would define an argument named arg1 for our function.\
What if you want to indicate what type the argument should be? Use the built-in syntax: ``` def example_function(arg1:str): ```.\
In the example we define that the earlier arg1 for our function is expected to be of string type. In the case your variable accepts several types, use braces:\
```def example_function(arg1:Union[str, int]):``` \
or the special ```|``` symbol:\
```def example_function(arg1:str | int): ```
Here arg1 can be of string type or integer type.\
If you want to indicate the output type of the function, here's the syntax for it: ```def example_function(arg1:Union[str, int])->float:```.\
Here we define the output to be of type float.\
While inside the function you're like inside of a regular python environment, imagine it to be a separate space, that can use the variables defined *globally*,\
however all of the variables defined within it are invisible for the global view and are thus *local*.\
To indicate the function for the regular function use the ```return``` keyword. For the generator use the ```yield``` keyword.\
What's the difference between the function and the generator? A function runs once it was called and then returns the output and all its variables cease\
to exist, while in the case of generator it *yields* values one by one until it exhausts all of the values, which is done via the ```next``` function:\
```next(some_generator())```, here we call the ```next``` function to get the next value from the generator.\
For now, we'll focus mainly on functions as they are *much* more often used than generators.\
Thus, a function that accepts some number then converts it to string would look something like that:

```python
def converter(number:int | float | complex)->str:
    return str(number)
```

Let's talk about decorators. What exactly are they? Well sometimes you need to take another function as an input and modify its output.\
A good example of that is when there's some function (or a group of functions), the output of which you want to change.\
So how exactly do they work? This is how a basic decorator is defined:

```python

def some_decorator(func):
    def wrapper():
        print('This is the space for the script that takes place before we call the target function')
        func()
        print('This is the space for the script that takes place after we call the target function')
    return wrapper
```

And this is how it's used:

````python
def some_decorator(func):
    def wrapper():
        #print('This is the space for the script that takes place before we call the target function')
        a_inside, b_inside = 4, 5
        c_inside, d_inside = func(a_inside, b_inside)
        print("Before adding 1:", c_inside)
        c_inside += 1 #add 1 to c_inside
        print("After adding 1: ", c_inside)
        #print('This is the space for the script that takes place after we call the target function')
        return c_inside, d_inside
    return wrapper


#define some function to apply the decorator to
@some_decorator
def sample_func(a, b):
    #print("I am just a function! I don't even have any output to return")
    print(a*b)
    return a * b, a+b

#Now when we'll call the function, it will be called with the decorator applied
val = sample_func()
print(val)
````

Honorable mentions: lambda functions.\
Lambda functions are also quite important in python, they are used to shorten the code for the functions or to call functions in the areas where you\
don't really want to bother with creating an entire new function (you will see these examples a lot more often later when working with the pandas library)\
They have a very simple syntax: using the keyword ```lambda``` and the names of the variables after it (note that the variable type specifications don't\
apply to the lambda functions) and then the function itself, like this: ```lambda x, y: x + y```. These functions can be created without having names,\
but if you want to give them names, just assign them to some variable, the name of which will be the name of the function:
```python
some_lambda_function = lambda x, y: x + y
```

4. Create a decorator timer_decorator that:\
Prints "Function started" before calling the function\
Calls the function\
Prints "Function completed" after the function finishes\
Apply it to a simple function that prints "Hello World"

````python
def timer_decorator(func):
    def wrapper():
        print('Function started')
        func()
        print('Function completed')
    return wrapper

@timer_decorator
def tmp():
    print('I am executing')

tmp()
````
