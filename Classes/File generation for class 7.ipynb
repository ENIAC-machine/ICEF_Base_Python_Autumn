```python
import numpy as np
import time

def solve_heat_equation_finite_difference(length=1.0, n_points=50, time_steps=100, alpha=0.01):
    """
    Solve 1D heat equation using finite difference method:
    ∂u/∂t = α * ∂²u/∂x²
    """
    # Spatial discretization
    dx = length / (n_points - 1)
    x = np.linspace(0, length, n_points)
    
    # Time discretization
    dt = 0.001  # time step
    
    # Stability condition check
    r = alpha * dt / (dx**2)
    if r > 0.5:
        dt = 0.5 * (dx**2) / alpha
        r = alpha * dt / (dx**2)
    
    # Initial condition (Gaussian pulse)
    u = np.exp(-((x - length/2)**2) / 0.05)
    
    # Boundary conditions (fixed ends)
    u[0] = 0.0
    u[-1] = 0.0
    
    # Time-stepping using explicit finite difference
    for step in range(time_steps):
        u_new = u.copy()
        
        # Update interior points
        for i in range(1, n_points - 1):
            u_new[i] = u[i] + r * (u[i+1] - 2*u[i] + u[i-1])
        
        # Apply boundary conditions
        u_new[0] = 0.0
        u_new[-1] = 0.0
        
        u = u_new
    
    return x, u

def generate_synthetic_data():
    """Generate synthetic data based on heat equation solution"""
    
    # Solve heat equation
    x, temperature = solve_heat_equation_finite_difference()
    
    # Calculate derived quantities
    max_temp = np.max(temperature)
    min_temp = np.min(temperature)
    avg_temp = np.mean(temperature)
    
    # Calculate temperature gradient (finite difference derivative)
    gradient = np.gradient(temperature, x)
    max_gradient = np.max(np.abs(gradient))
    
    # Calculate heat flux (Fourier's law: q = -k * dT/dx)
    k = 0.5  # thermal conductivity
    heat_flux = -k * gradient
    avg_heat_flux = np.mean(np.abs(heat_flux))
    
    # Calculate thermal energy (approximation: ∫T dx ≈ sum(T)*dx)
    dx = x[1] - x[0]
    thermal_energy = np.sum(temperature) * dx
    
    return {
        'max_temperature': max_temp,
        'avg_temperature': avg_temp,
        'max_gradient': max_gradient,
        'thermal_energy': thermal_energy
    }

def write_to_log_file(filename='logs.txt', num_entries=50, interval=0.1):
    """
    Write synthetic data to log file
    
    Args:
        filename: Name of the output file
        num_entries: Number of data entries to generate
        interval: Time between entries (seconds)
    """
    
    with open(filename, 'w') as f:
        for i in range(num_entries):
            # Generate synthetic data
            data = generate_synthetic_data()
            
            # Add some random variation to simulate real measurements
            noise_factor = 0.01 * np.random.randn()
            
            # Format: variable_1=..., variable_2=...
            log_line = (f"max_temperature={data['max_temperature'] + noise_factor:.6f}, "
                       f"avg_temperature={data['avg_temperature'] + noise_factor*0.5:.6f}, "
                       f"max_gradient={data['max_gradient'] + abs(noise_factor)*0.1:.6f}, "
                       f"thermal_energy={data['thermal_energy'] + noise_factor*2:.6f}\n")
            
            f.write(log_line)
            
            # Print progress
            if i % 10 == 0:
                print(f"Generated entry {i+1}/{num_entries}")
            
            # Wait before next entry
            time.sleep(interval)
    
    print(f"\nData written to {filename}")
    print(f"Generated {num_entries} entries with {interval}s interval")

def read_and_display_log_file(filename='logs.txt', num_lines=5):
    """Read and display first few lines from log file"""
    print(f"\nFirst {num_lines} lines from {filename}:")
    print("-" * 80)
    
    with open(filename, 'r') as f:
        lines = f.readlines()
        for i, line in enumerate(lines[:num_lines]):
            print(f"Line {i+1}: {line.strip()}")
    
    print("-" * 80)
    print(f"Total entries: {len(lines)}")

def main():
    """Main function to run the data generation"""
    print("Generating synthetic signal data from heat equation simulation...")
    print("Using finite-difference method for 1D heat equation: ∂u/∂t = α * ∂²u/∂x²")
    print("\nVariables being logged:")
    print("1. max_temperature: Maximum temperature in the domain")
    print("2. avg_temperature: Average temperature across the domain")
    print("3. max_gradient: Maximum temperature gradient")
    print("4. thermal_energy: Total thermal energy in the system")
    
    # Generate and write data
    write_to_log_file('logs.txt', num_entries=30, interval=0.05)
    
    # Display sample of the data
    read_and_display_log_file('logs.txt', num_lines=5)
    
    # Optional: Plot the data
    plot_data_option = input("\nWould you like to visualize the last temperature profile? (y/n): ")
    if plot_data_option.lower() == 'y':
        plot_temperature_profile()

def plot_temperature_profile():
    """Optional: Plot the temperature profile from the last simulation"""
    try:
        import matplotlib.pyplot as plt
        
        x, temperature = solve_heat_equation_finite_difference()
        
        plt.figure(figsize=(10, 6))
        plt.plot(x, temperature, 'b-', linewidth=2, label='Temperature')
        plt.xlabel('Position (x)', fontsize=12)
        plt.ylabel('Temperature (T)', fontsize=12)
        plt.title('Temperature Profile from Finite-Difference Solution of Heat Equation', fontsize=14)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.savefig('temperature_profile.png', dpi=100)
        plt.show()
        print("Temperature profile saved as 'temperature_profile.png'")
    except ImportError:
        print("Matplotlib not installed. Skipping plot generation.")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import time
import random
import string
from datetime import datetime

def solve_heat_equation_finite_difference(length=1.0, n_points=50, time_steps=100, alpha=0.01):
    """
    Solve 1D heat equation using finite difference method:
    ∂u/∂t = α * ∂²u/∂x²
    """
    # Spatial discretization
    dx = length / (n_points - 1)
    x = np.linspace(0, length, n_points)
    
    # Time discretization
    dt = 0.001
    
    # Stability condition check
    r = alpha * dt / (dx**2)
    if r > 0.5:
        dt = 0.5 * (dx**2) / alpha
        r = alpha * dt / (dx**2)
    
    # Initial condition (Gaussian pulse)
    u = np.exp(-((x - length/2)**2) / 0.05)
    
    # Boundary conditions
    u[0] = 0.0
    u[-1] = 0.0
    
    # Time-stepping
    for step in range(time_steps):
        u_new = u.copy()
        for i in range(1, n_points - 1):
            u_new[i] = u[i] + r * (u[i+1] - 2*u[i] + u[i-1])
        u_new[0] = 0.0
        u_new[-1] = 0.0
        u = u_new
    
    return x, u

def generate_corrupted_value(original_value, corruption_type="random_string"):
    """
    Generate a corrupted value instead of the actual number
    """
    if corruption_type == "random_string":
        # Generate a random alphanumeric string
        length = random.randint(3, 8)
        return ''.join(random.choices(string.ascii_letters + string.digits, k=length))
    
    elif corruption_type == "special_chars":
        # Generate string with special characters
        special_chars = "!@#$%^&*()_+-=[]{}|;:,.<>?"
        length = random.randint(4, 10)
        chars = random.choices(string.ascii_letters + string.digits + special_chars, k=length)
        return ''.join(chars)
    
    elif corruption_type == "hex_value":
        # Generate a hex-like string
        hex_digits = "0123456789ABCDEF"
        return "0x" + ''.join(random.choices(hex_digits, k=random.randint(4, 8)))
    
    elif corruption_type == "date_format":
        # Generate a date-like string
        year = random.randint(2000, 2024)
        month = random.randint(1, 12)
        day = random.randint(1, 28)
        return f"{year:04d}-{month:02d}-{day:02d}"
    
    elif corruption_type == "null_like":
        # Generate null-like values
        null_options = ["NULL", "null", "None", "NaN", "N/A", "undefined"]
        return random.choice(null_options)
    
    else:
        # Mix of letters and numbers
        length = random.randint(5, 12)
        return ''.join(random.choices(string.ascii_letters + string.digits, k=length))

def generate_synthetic_data():
    """Generate synthetic data based on heat equation solution"""
    
    x, temperature = solve_heat_equation_finite_difference()
    
    # Calculate derived quantities
    max_temp = np.max(temperature)
    min_temp = np.min(temperature)
    avg_temp = np.mean(temperature)
    
    # Calculate temperature gradient
    gradient = np.gradient(temperature, x)
    max_gradient = np.max(np.abs(gradient))
    
    # Calculate heat flux
    k = 0.5 + 0.1 * np.random.randn()  # Add some variation to k
    heat_flux = -k * gradient
    avg_heat_flux = np.mean(np.abs(heat_flux))
    
    # Calculate thermal energy
    dx = x[1] - x[0]
    thermal_energy = np.sum(temperature) * dx
    
    # Add natural variation to simulate different conditions
    variation = 1.0 + 0.05 * np.random.randn()
    
    return {
        'max_temperature': max_temp * variation,
        'avg_temperature': avg_temp * variation,
        'max_gradient': max_gradient * variation,
        'thermal_energy': thermal_energy * variation,
        'timestamp': datetime.now().strftime("%H:%M:%S.%f")[:-3]
    }

def corrupt_entry(data, corruption_probability=0.2, corruption_pattern="random"):
    """
    Corrupt an entry with given probability
    
    Args:
        data: Dictionary of data values
        corruption_probability: Probability of corruption (0 to 1)
        corruption_pattern: How to corrupt the data:
            - "random": Randomly replace some values
            - "all": Corrupt all values in entry
            - "specific": Corrupt specific variables
    """
    if random.random() > corruption_probability:
        return data  # No corruption
    
    corrupted_data = data.copy()
    variables = ['max_temperature', 'avg_temperature', 'max_gradient', 'thermal_energy']
    
    if corruption_pattern == "random":
        # Corrupt random number of variables (1 to 3)
        num_to_corrupt = random.randint(1, 3)
        vars_to_corrupt = random.sample(variables, num_to_corrupt)
        
        for var in vars_to_corrupt:
            corruption_type = random.choice(["random_string", "special_chars", "hex_value", "date_format", "null_like"])
            corrupted_data[var] = generate_corrupted_value(data[var], corruption_type)
    
    elif corruption_pattern == "all":
        # Corrupt all variables
        for var in variables:
            corruption_type = random.choice(["random_string", "special_chars", "hex_value", "date_format", "null_like"])
            corrupted_data[var] = generate_corrupted_value(data[var], corruption_type)
    
    elif corruption_pattern == "specific":
        # Corrupt based on position in string
        position = random.choice(["first_half", "second_half", "alternating"])
        
        if position == "first_half":
            for var in variables[:2]:  # First two variables
                corruption_type = random.choice(["random_string", "hex_value"])
                corrupted_data[var] = generate_corrupted_value(data[var], corruption_type)
        
        elif position == "second_half":
            for var in variables[2:]:  # Last two variables
                corruption_type = random.choice(["special_chars", "null_like"])
                corrupted_data[var] = generate_corrupted_value(data[var], corruption_type)
        
        else:  # alternating
            for i, var in enumerate(variables):
                if i % 2 == 0:
                    corruption_type = random.choice(["random_string", "date_format"])
                    corrupted_data[var] = generate_corrupted_value(data[var], corruption_type)
    
    # Add corruption marker to timestamp
    corrupted_data['timestamp'] = corrupted_data['timestamp']
    
    return corrupted_data

def format_log_line(data, include_timestamp=True, format_type="standard"):
    """
    Format a log line from data dictionary
    
    Args:
        data: Dictionary with data values
        include_timestamp: Whether to include timestamp
        format_type: "standard" or "mixed" formatting
    """
    if format_type == "mixed":
        # Occasionally mix up the order of variables
        if random.random() < 0.1:
            variables = ['max_gradient', 'thermal_energy', 'max_temperature', 'avg_temperature']
            line_parts = [f"{var}={data[var]}" for var in variables]
        else:
            variables = ['max_temperature', 'avg_temperature', 'max_gradient', 'thermal_energy']
            line_parts = [f"{var}={data[var]}" for var in variables]
    else:
        variables = ['max_temperature', 'avg_temperature', 'max_gradient', 'thermal_energy']
        line_parts = [f"{var}={data[var]}" for var in variables]
    
    if include_timestamp:
        line_parts.append(f"timestamp={data['timestamp']}")
    
    return ", ".join(line_parts)

def write_to_log_file(filename='logs.txt', num_entries=50, interval=0.1, 
                      corruption_probability=0.2, corruption_pattern="random"):
    """
    Write synthetic data to log file with corruption
    
    Args:
        corruption_probability: Probability of an entry being corrupted (0 to 1)
        corruption_pattern: How to corrupt entries ("random", "all", "specific")
    """
    
    stats = {'total': 0, 'corrupted': 0, 'clean': 0}
    
    with open(filename, 'w') as f:
        for i in range(num_entries):
            # Generate clean synthetic data
            clean_data = generate_synthetic_data()
            stats['total'] += 1
            
            # Decide if we should corrupt this entry
            should_corrupt = random.random() < corruption_probability
            
            if should_corrupt:
                data = corrupt_entry(clean_data, corruption_probability=1.0, 
                                    corruption_pattern=corruption_pattern)
                stats['corrupted'] += 1
            else:
                data = clean_data
                stats['clean'] += 1
            
            # Format the log line (occasionally mix up order for variety)
            format_type = "mixed" if random.random() < 0.3 else "standard"
            log_line = format_log_line(data, include_timestamp=True, format_type=format_type)
            
            f.write(log_line + "\n")
            
            # Print progress with corruption indication
            if i % 10 == 0:
                status = " [CORRUPTED]" if should_corrupt else ""
                print(f"Generated entry {i+1}/{num_entries}{status}")
            
            # Wait before next entry
            time.sleep(interval)
    
    print(f"\nData written to {filename}")
    print(f"Generated {num_entries} entries with {interval}s interval")
    print(f"Clean entries: {stats['clean']}, Corrupted entries: {stats['corrupted']} " +
          f"({stats['corrupted']/stats['total']*100:.1f}% corrupted)")
    
    return stats

def read_and_analyze_log_file(filename='logs.txt', num_lines=10):
    """Read and analyze log file, identifying corrupted entries"""
    print(f"\nAnalyzing {filename}:")
    print("-" * 80)
    
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    # Pattern to detect numbers (including scientific notation)
    number_pattern = r'[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?'
    
    corrupted_entries = []
    clean_entries = []
    
    for i, line in enumerate(lines[:min(num_lines, len(lines))]):
        # Check if line contains non-numeric values where numbers are expected
        parts = line.strip().split(', ')
        
        # Look for variable=value patterns
        has_corruption = False
        corrupted_vars = []
        
        for part in parts:
            if '=' in part:
                var, value = part.split('=', 1)
                # Check if it's one of our numeric variables
                if var in ['max_temperature', 'avg_temperature', 'max_gradient', 'thermal_energy']:
                    try:
                        float(value)
                    except ValueError:
                        has_corruption = True
                        corrupted_vars.append(var)
        
        if has_corruption or '[CORRUPTED]' in line:
            status = "CORRUPTED"
            if corrupted_vars:
                status += f" (vars: {', '.join(corrupted_vars)})"
            corrupted_entries.append(i+1)
        else:
            status = "CLEAN"
            clean_entries.append(i+1)
        
        print(f"Line {i+1} [{status}]: {line.strip()}")
    
    print("-" * 80)
    print(f"Total entries: {len(lines)}")
    print(f"Sample analysis: {len(clean_entries)} clean, {len(corrupted_entries)} corrupted in first {num_lines} lines")
    
    if corrupted_entries:
        print(f"Corrupted entry numbers in sample: {corrupted_entries}")
    
    return lines

def generate_corruption_report(filename='logs.txt'):
    """Generate a detailed report of corruption patterns"""
    print(f"\nGenerating corruption report for {filename}:")
    print("=" * 60)
    
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    corruption_types = {
        'string': 0,
        'special_chars': 0,
        'hex': 0,
        'date': 0,
        'null': 0,
        'mixed': 0
    }
    
    corruption_by_variable = {
        'max_temperature': 0,
        'avg_temperature': 0,
        'max_gradient': 0,
        'thermal_energy': 0
    }
    
    for i, line in enumerate(lines):
        parts = line.strip().split(', ')
        for part in parts:
            if '=' in part:
                var, value = part.split('=', 1)
                if var in corruption_by_variable:
                    try:
                        float(value)
                    except ValueError:
                        # This is a corrupted value
                        corruption_by_variable[var] += 1
                        
                        # Classify corruption type
                        if '0x' in value and all(c in '0123456789ABCDEFabcdef' for c in value[2:]):
                            corruption_types['hex'] += 1
                        elif any(c in '!@#$%^&*()_+-=[]{}|;:,.<>?' for c in value):
                            corruption_types['special_chars'] += 1
                        elif value in ['NULL', 'null', 'None', 'NaN', 'N/A', 'undefined']:
                            corruption_types['null'] += 1
                        elif '-' in value and len(value) == 10:  # Likely date format
                            corruption_types['date'] += 1
                        elif value.isalnum():
                            corruption_types['string'] += 1
                        else:
                            corruption_types['mixed'] += 1
    
    print("\nCorruption Statistics:")
    print("-" * 40)
    total_corruptions = sum(corruption_by_variable.values())
    if total_corruptions > 0:
        print(f"Total corruptions detected: {total_corruptions}")
        
        print("\nCorruption by variable:")
        for var, count in corruption_by_variable.items():
            if count > 0:
                print(f"  {var}: {count} ({count/total_corruptions*100:.1f}%)")
        
        print("\nCorruption types:")
        for ctype, count in corruption_types.items():
            if count > 0:
                print(f"  {ctype}: {count}")
    
    else:
        print("No corruptions detected in the file")
    
    print("=" * 60)

def main():
    """Main function to run the data generation with corruption"""
    print("Generating synthetic signal data with corruption...")
    print("Using finite-difference method for 1D heat equation: ∂u/∂t = α * ∂²u/∂x²")
    print("\nVariables being logged:")
    print("1. max_temperature: Maximum temperature in the domain")
    print("2. avg_temperature: Average temperature across the domain")
    print("3. max_gradient: Maximum temperature gradient")
    print("4. thermal_energy: Total thermal energy in the system")
    
    # Get user input for corruption parameters
    try:
        corruption_prob = float(input("\nEnter corruption probability (0.0 to 1.0, default 0.2): ") or "0.2")
        num_entries = int(input("Enter number of entries to generate (default 50): ") or "50")
        print("\nCorruption patterns available:")
        print("  random: Randomly corrupt variables in an entry")
        print("  all: Corrupt all variables in an entry")
        print("  specific: Corrupt specific patterns of variables")
        corruption_pattern = input("Enter corruption pattern (default random): ") or "random"
    except ValueError:
        print("Invalid input, using defaults")
        corruption_prob = 0.2
        num_entries = 50
        corruption_pattern = "random"
    
    # Generate and write data with corruption
    stats = write_to_log_file('logs.txt', 
                             num_entries=num_entries,
                             interval=0.05,
                             corruption_probability=corruption_prob,
                             corruption_pattern=corruption_pattern)
    
    # Display sample of the data
    lines = read_and_analyze_log_file('logs.txt', num_lines=10)
    
    # Generate corruption report
    generate_corruption_report('logs.txt')
    
    # Option to show examples of corruption
    show_examples = input("\nShow examples of different corruption types? (y/n): ")
    if show_examples.lower() == 'y':
        print("\nExamples of corruption types:")
        print("-" * 50)
        test_value = 12.3456
        for ctype in ["random_string", "special_chars", "hex_value", "date_format", "null_like"]:
            corrupted = generate_corrupted_value(test_value, ctype)
            print(f"{ctype}: {test_value:.4f} -> {corrupted}")
    
    # Option to create a clean version
    create_clean = input("\nCreate a clean version without corruption? (y/n): ")
    if create_clean.lower() == 'y':
        write_to_log_file('logs_clean.txt', 
                         num_entries=num_entries,
                         interval=0.01,
                         corruption_probability=0.0)
        print("Clean version saved as 'logs_clean.txt'")

if __name__ == "__main__":
    main()
```
