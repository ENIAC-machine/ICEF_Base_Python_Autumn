# Return keyword

```python
num = 10 #define variable num that's set to 10

def example_function(some_number:int=0): #here we define a function named "example_function"
    '''
    This function modifies variable some_number in some way
    
    return keyword is used for function's standard output
    '''

    some_number *= 5
    some_number /= 2
    num = some_number
    print("num variavble inside the function: ", num)
    return num #equivalent to "return some_number"

num = example_function(num)
print("num variable outside the function: ", num)
```
