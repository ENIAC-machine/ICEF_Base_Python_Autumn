# Python Class

There are times when you need to store some data in one place **and** have a set of methods that's applicable to that data specifically\
On the everyday scale ```Objects``` provide convenience and lightening of code, let's now examine python ```Class```, which is the python implementation of the ```Object```  

````python

#you initiate a python class via the class keyword, here 'Hey' is the name of the class 
class Hey():

    #__init__ here is what is called a ✨magic✨ method, we'll touch on what those are a bit later
    #just know that __init__ is automatically executed when the class is initialized (called)
    def __init__(self) -> None:
        '''
        Class initialization

        Inputs:
            self - all the instances (variables, functions, etc.) attached to this class
        '''
        
        print("Hello, dude!")

#Let's now try calling the class
hey = Hey()
````

What if we want to pass some arguments to the class and them use them with some class-specific functions?\
It's made for it after all\
Let's do that solving a toy problem of calculating the area of a circle given its radius

````python
class Circle(): #note that you don't pass arguments to the class when you initialize it

    #this variable is shared by all the instances of the class 'Circle', we could (impropelry) call it the class global variable
    entity = 'circle'
    
    #initalize the class with the requirement to pass a variable to it
    def __init__(self, radius: float = None) -> None:
        '''

        Initialize the class

        Inputs:
            radius: float - radius of a circle
            
        '''
        #This variable is specific to each instance of the class 'Circle'
        self.r = radius

    def calculate_area(self, save: bool = False, out: bool = True)-> float | None:
        '''
        Calculate the are of the circle

        Inputs:
            self
            
            save: bool - whether to save the value to the class variable self.area
            
            out: bool - whether to output the value

            
        Outputs:
            area : float - area of the circle

        '''

        #Calculate the area as A = pi*r^2
        area = 3.14*self.r**2

        #if save is True, save the value to the class
        if save:
            self.area = area

        #if output is True, return the area, otherwise return None
        if out:
            return area
        

#Let's now create two instances of the class 'Circle' and see what are their similarities and differences

big = Circle(10) #circle with radius 10
small = Circle(2) #circle with radius 2
````

````python
#Let's first see the variables they share:
print(f'{big.entity}, {small.entity}')
````

````python
#Now let's look at the Class-specific variables
print(f'Radius of the big circle={big.r}, radius of the small circle={small.r}')
````

Now let's try calculating the area of the circles

````python
#In case we want to only save it to the variable outside of the scope of the class
area_big = big.calculate_area(save=False, out=True)
area_small = small.calculate_area(save=False, out=True)
print(f'{area_big}, {area_small}', end= '\n\n\n')

#In case we want to only save it as a class-specific variable
big.calculate_area(save=True, out=False)
small.calculate_area(save=True, out=False)
print(f'{big.area}, {small.area}', end= '\n\n\n')

#In case we want to both save it as a class-specific variable and the variable outside of the scope of the class
area_big = big.calculate_area(save=True, out=True)
area_small = small.calculate_area(save=True, out=True)
print(f'{area_big}, {big.area} | {area_small}, {small.area}')
````

# ✨Magic✨ and static methods

## ✨Magic✨ methods

We've already applied a whole 2 methods to our class, which are ```__init__``` and ```calculate_area```, yet one can notice a striking difference between them:\
```__init__``` executes automatically, while ```calculate_area``` needed to be specifically called, why?\
Let's first examine the group of function, to which the ```__init__``` belongs- these are called the ✨magic✨ functions:

These are the set of functions with the underlines on both sides that python executes automatically in some specific circumstances. You already know that ```__init__``` executes upon the initialization of the class, let's see other examples

```__repr__```

````python

"""
__repr__ is a ✨magic✨ function that is responsible for the representation of the class a.k.a. the output string,
which is displayed when you are outputing the class. For example the __repr__ method for the class of an integer variable in python is the number itself

"""

#Let's first see how the representation will look without the __repr__ method defined
class Question():

    def __init__(self, query: str) -> None:
        self.q = query

a = Question('Why are we here?')
print(f"If __repr__ isn't defined: {a}")
````

That doesn't look pretty, let's redefine it

````python
class Question():

    def __init__(self, query: str) -> None:
        self.q = query

    #Let's make the representation of the Question class the query itself
    def __repr__(self) -> str:
        return self.q

a = Question('Why are we here?')
print(f"If __repr__ is defined: {a}")
````

```__bool__```

You have already encountered the code of type:

```python

num = 15
if num:
    print('Hey!')

```

In instances like this python uses ```__bool__``` method to convert the non-boolean class into a boolean variable and that's what we shall implement

````python
#Let's define a sample class that would store an array, this way converting to boolean is simple - if the stored array is empty-> False otherwise True
class Records():

    def __init__(self) -> None:
        
        self.recs = []

    def __bool__(self) -> bool:
        if len(self.recs) == 0:
            return False
        else:
            return True

    def add_rec(self, rec) -> None:

        self.recs.append(rec)

storage = Records()

if storage:
    print("This string shouldn't output ")

print(f'Current boolean value for the variable storage: {bool(storage)}', end='\n\n\n')

storage.add_rec(12)


if storage:
    print('This string should output')
    
print(f'Current boolean value for the variable storage: {bool(storage)}')
````

```__str__```

This ✨magic✨ method is responsible for the value during the conversion to the ```str``` variable. See example below

````python
class Complex():

    def __init__(self, real: float, im: float) -> None:
        self.real = real
        self.im = im

    def __str__(self) -> str:
        return f'{self.real}{self.im}i'


num = str(Complex(10.12, -5.6))
print(num)
````

## Static methods

Sometimes we don't want to include the ```self``` argument into the arguments pool of our class methods, that's when we use the ```@staticmethod``` decorator\
This is practical, when you want flexibility for some of your methods to be able to use them both within the class and with some outside values\
It's used when you want to create a function, which wouldn't depend on the class specific data, let's view an example

````python
class Square():

    def __init__(self) -> None:
        return

    
    @staticmethod
    def calc_area(pts: tuple[tuple[float]], save: bool= False)->float:
        '''
        
        Calculate the area of an arbitrary 4-edge polygon 
        
        As input this function takes a tuple of the form ((x1, y1), (x2, y2), (x3, y3), (x4, y4)),
        where the positions of the points are illustrated below
        
        (x4, y4)--------(x3, y3)
            |               | 
            |               | 
            |               | 
            |               | 
        (x1, y1)--------(x2, y2)
        
        '''
        assert len(pts) == 4, f"Wrong input type, expected length 4, got length of {len(pts)}"

        area = .5 * abs(pts[0][0]*pts[1][1] + pts[1][0]*pts[2][1] + pts[2][0]*pts[3][1] + pts[3][0]*pts[0][1]\
                        - pts[0][1]*pts[1][0] - pts[1][1]*pts[2][0] - pts[2][1]*pts[3][0] - pts[3][1]*pts[0][0]
                       )
        
        if save:
            self.area = area
        
        return area

sq = Square()
#As you can see we can use arbitrary values, disconnected from our class to use this function
area = sq.calc_area(((0, 3), (2, 4), (2, 5), (0,5)))
print(f"The area of the polygon is {area:.2f}") #Check out the cool format specificator I used (:.2f)
````
