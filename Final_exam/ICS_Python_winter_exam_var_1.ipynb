    [[0., 1.],

Iteration 1: the snow falls down so our values ``[0, 1]`` are now one row below and a new couple of values is\ generated - ``[1, 1]``. They are put in the first row. Thus, the state at iteration 1 is:

    [[1., 1.],
     [0., 1.],
     [0., 0.]]

Iteration 2: the snow falls again, the story is the same as last time. Current state:

    [[0., 0.],
     [1., 1.],
     [0., 1.]]

Iteration 3: as the snow falls, it accumulates on the ground. So we add the values from the row that\
is second to last with the values from the last row to get the final state (of course we also generated new snow):

    [[0., 1.],
     [0., 0.],
     [1., 2.]]
