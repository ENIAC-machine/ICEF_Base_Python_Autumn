```python
import numpy as np

def generate_corrupted_motion_data(output_times=[10, 11, 20, 21, 30], 
                                  corruption_prob=0.0, 
                                  filename=None,
                                  num_entries=None,
                                  initial_velocity=2.0):
    """
    Generate motion simulation data with optional corruption.
    
    Parameters:
    -----------
    output_times : list
        Times at which to output data points
    corruption_prob : float (0.0 to 1.0)
        Probability of corrupting each data entry
    filename : str or None
        Output filename (None generates a unique name)
    num_entries : int or None
        Number of entries to generate (overrides output_times if provided)
    initial_velocity : float
        Initial velocity value for simulation
    
    Returns:
    --------
    str : Output filename
    """
    
    # Validate input arguments with assert statements
    # Check output_times when num_entries is not provided
    if num_entries is None:
        assert isinstance(output_times, list), "output_times must be a list"
        assert len(output_times) > 0, "output_times must not be empty"
        assert all(isinstance(t, (int, np.integer)) for t in output_times), \
            "All elements in output_times must be integers"
        assert all(t >= 0 for t in output_times), \
            "All time values in output_times must be non-negative"
        assert len(output_times) == len(set(output_times)), \
            "output_times must contain unique values (no duplicates)"
    
    # Check num_entries when provided
    if num_entries is not None:
        assert isinstance(num_entries, (int, np.integer)), \
            "num_entries must be an integer"
        assert num_entries > 0, "num_entries must be greater than 0"
    
    # Check corruption_prob
    assert isinstance(corruption_prob, (int, float, np.number)), \
        "corruption_prob must be a number"
    assert 0.0 <= corruption_prob <= 1.0, \
        "corruption_prob must be between 0.0 and 1.0 inclusive"
    
    # Check filename
    if filename is not None:
        assert isinstance(filename, str), "filename must be a string"
        assert len(filename) > 0, "filename must not be empty"
        assert "." in filename, "filename should have an extension"
    
    # Check initial_velocity
    assert isinstance(initial_velocity, (int, float, np.number)), \
        "initial_velocity must be a number"
    
    # Generate unique filename if not provided
    if filename is None:
        uid = str(np.random.randint(1000000, 9999999))
        filename = f"motion_{uid}.txt"
    
    # Determine time points
    if num_entries is not None:
        t = np.arange(num_entries)
        output_times = list(range(num_entries))
    else:
        t = np.arange(max(output_times) + 1)
    
    # Random waves for unique acceleration
    waves = np.array([
        np.random.uniform(0.1, 0.3) * np.sin(t * np.random.uniform(0.1, 0.5)),
        np.random.uniform(0.05, 0.2) * np.cos(t * np.random.uniform(0.5, 1.0)),
        -np.random.uniform(0.03, 0.1) * np.sin(t * np.random.uniform(1.0, 2.0))
    ])
    
    # Calculations
    a = waves.sum(axis=0)
    v = initial_velocity + a.cumsum()
    d = v.cumsum()
    
    # Function to potentially corrupt a value
    def corrupt_value(value):
        if np.random.random() < corruption_prob:
            # Generate random corruption (mix of symbols and letters)
            corruptions = ['#ERR#', '@NaN@', '*INF*', '!COR!', '???.??', 'XXXXX']
            return np.random.choice(corruptions)
        return f"{value:.1f}"
    
    # Write to file with potential corruption
    with open(filename, 'w') as f:
        # Write data directly without header
        for i in output_times:
            # Get values and potentially corrupt them
            time_val = i
            speed_val = corrupt_value(v[i])
            dist_val = corrupt_value(d[i])
            
            # Write line to file
            f.write(f"t={time_val}, speed={speed_val}, cum_dist={dist_val}\n")
            
            # Print to console (with corruption indication)
            print(f"t={time_val}, speed={speed_val}, cum_dist={dist_val}")
    
    print(f"\nSaved {len(output_times)} entries to {filename}")
    return filename

generate_corrupted_motion_data(corruption_prob=0.1, num_entries=100)
```
