# Problem 2 (50 pts). 1D Cellular Automata

In this task you will create a 1D version of the famous game of life.\
Write a function, which will do the following:

1. Initialize the array of size ```N``` of 1's and 0's of user's choosing. Note that user can either pass the array in the ```arr``` argument 
(but it must consist only of 0's and 1's) or they can pass just the size of the array in the ```N``` argument. Both arguments can be non-trivial values
if and only if len(```arr```) equals ```N```. The user also must be able to provide the seed in case no ```arr``` is passed
2. Display the game rules, they can come by default (see the default rules below) or be set by the user in the form of a dictionary in the ```rules``` argument. ```rules``` denote the value of the cell given the values of the neighbors, for example '010' : '0' says that if the value of this cell is 1
and the neighbors are 0's then we set it to 0, another example is '011' : '1', which means that if the cell is 1 and cells to the right and the left of
it are 1 and 0 respectively, then the cell's value transforms into 1. For the corner values you can use the cut version of the rule, for example for the
first cell you may interpret '10' as '010', '00' as '000' and etc.
3. Perform the simulation ```T``` times, where ```T``` is a required argument
4. Output the array after the simulation

Constraints:
1. You are allowed 1 loop for this exercise
2. Available libraries: ```numpy```
3. ```arr``` consists **only** of 1's and 0's of integer format
4. ```N``` is a positive integer as is ```T```
5. ```rules``` contains exactly **8** keys

Example:

N = 5\
T = 2\
rules = {'000' : '0', '001' : 1, '010' : '0', '011' : '1', '100' : '1', '101' : '0', '110' : '1', '111' : 0}\
arr = [1, 0, 1, 1 ,1]

On the first iteration:\
[1, 0, 1, 1, 1] -> [0, 0, 1, 0, 1]

On the second iteration:\
[0, 0, 1, 0, 1] -> [0, 1, 0, 0, 0]

Name the function ```CA```

````python
def CA(T: int,
       N: int = None,
       arr: np.ndarray | list = None,
       rules: dict = {'000' : '0', '001' : 1, '010' : '0', '011' : '1', '100' : '1', '101' : '0', '110' : '1', '111' : 0},
       seed: int= None) -> np.ndarray:
    
    '''
    
    Game of life simulation


    Inputs:
        N: int - length of array
        arr: int - array itself. If passed with N, must satisfy condition N == len(arr)
        rules: dict - rules, according to which the evolution will be
        T: int - number of periods of simulation estimation
        seed: int - random seed

    Outputs:
        out: np.ndarray - the array after the simulation
        
    '''
    assert (isinstance(N, int) and N > 0) or N is None, "Wrong type of input for N"
    assert isinstance(T, int) and T > 0, 'Wrong type of input for T'
    assert isinstance(seed, int) or seed is None, "Wrong type of input for seed"
    assert isinstance(arr, list) or isinstance(arr, np.ndarray) or arr is None, "Wrong type of input for arr"
    assert isinstance(rules, dict), 'Wrong type of input for variable rules'

    assert set(rules.keys()) == {'000', '001', '010', '011', '100', '101', '110', '111'}, 'Wrong type of input for rules, not enough or too many'
    
    assert N or arr, "You must pass at least one of the variables- N, arr"

    if seed:
        np.random.seed(seed)
    
    if arr and not N:
        assert set(arr) == {0, 1}, "Wrong contents for arr"
        arr = np.array(arr)
        N = len(arr) #arr.shape[0]

    elif N and array is None:
        arr = np.random.randint(0, 2, N)
    
    if N and arr is not None:
        assert N == len(arr), "Wrong input for N or/and arr"

    print(f'Rules: {rules}')
    
    
    arr = arr.reshape(-1, 1)
    arr_idxs = np.arange(N).reshape(-1, 1)
    arr = np.concatenate([arr, arr_idxs], axis=1)
    print(arr)


    def func(x: np.ndarray) -> np.ndarray:
        val = x[0]
        idx = x[1]
        
        #get the neighbors of our value
        vals = arr[max(idx-1, 0):idx+2, 0]
        
        if len(vals) < 3 and idx == 0:
            vals = np.concatenate([[0], vals], axis=0)
        elif len(vals) < 3:
            vals = np.concatenate([vals, [0]], axis=0)

        vals = ''.join(vals.astype(str))
        val = int(rules[vals])
        return [val, idx] #[int(rules[vals]), idx]
    
    #simulation
    for t in range(T):
        arr = np.apply_along_axis(arr=arr, func1d= func, axis=1)

    return arr[:, 0]

CA(arr= [1, 0, 1, 1 ,1], T=2)
````
