# Problem 1 (20 pts). Cleaning duty

In this task you must write a function, which takes as input the path to some text file, note that the path may or may not include the ```.txt``` extension.\
The file contents are organized in the following pattern:

```
t=10, speed=1.9, cum_dist=8.0
t=11, speed=2.1, cum_dist=9.9

```

However, some entries in the file have corrupted values. Your function must detect these entries and discard them, rewriting the file so that it doesn't
contain such lines anymore while adding the acceleration to the listed values (add it to the end of the line in the same comma-separated format).
Note that your function must output an error message (created by you) in case the file path is invalid (there's not file with such path) and return ```None```. You may use the formula below to compute acceleration:

$$ \LARGE a_{t_i} = \frac{v_{t_i} - v_{t_{i-1}}}{t_i - t_{i-1}}$$

Example:
You are given entries 
```

t=10.5, speed=1.9, cum_dist=8.0
t=11.3, speed=2.1, cum_dist=9.AH
t=11.9, speed=2.0, cum_dist=10.7

```
Entry for t=11.3 is corrupted, so we exclude it, now to calculate the acceleration for t=11.9 we do:

$$ \Large a_{11.9} = \frac{2.0 - 1.9}{11.9-10.5}$$

New contents of the file:
```

t=10.5, speed=1.9, cum_dist=8.0, acceleration=None
t=11.9, speed=2.0, cum_dist=10.7, acceleration=0.07142857142857141

```

It is enough to be precise up to 3 digits after comma

Name the function ```cleanup```

Constraints:
1. No libraries can be used here except for the ```typing``` library

````python
def cleanup(path: str) -> None:
    '''

    Cleans up the file along the provided path and adds acceleration value where possible

    Inputs:
        path: str - path to the .txt file of the neccessary format 
    '''

    try:

        with open(path, 'r') as file:
            contents = file.readlines()

    except:
        print("File wasn't found (or other problems with opening the file)")
        return None

    ts, vs, cds = [], [], []

    for line in contents:
        line = line.strip('\n') #
        # print(line)
        # print(line.split('='))
        _, t, v, cd = line.split('=')
        # print(f'{t}| {v}| {cd}')
        t, v = t.split(',')[0], v.split(',')[0]
        # print(t, v, cd)
        try:
            ts.append(float(t))
            vs.append(float(v))
            cds.append(float(cd))
        except:
            pass

    accs = [None]

    # for idx, t, v in enumerate(zip(ts[1:], vs[1:])):
    #     accs.append(  (v - vs[idx-1]) / (t - ts[idx-1])  )

    for idx in range(1, len(vs)):
        accs.append( round( (vs[idx] - vs[idx-1]) / (ts[idx] - ts[idx-1]), 3) )

    # print(accs)

    # print(f'{ts}\n\n{vs}\n\n{cds}\n\n{accs}')

    with open(path, 'w') as file_new:
        for t, v, cd, acc in zip(ts, vs, cds, accs):
            file_new.write(f't={t}, speed={v}, cum_dist={cd}, acceleration={acc}\n')

    return None


cleanup('motion_3607981.txt')
````

Example:

arr = [1, 0, 1, 1 ,1]

On the first iteration:\
[1, 0, 1, 1, 1] -> [0, 0, 1, 0, 1]

On the second iteration:\
[0, 0, 1, 0, 1] -> [0, 1, 0, 0, 0]

Name the function ```CA```

````python

def CA(T: int, N: int=None, arr: list=None,
       rules: dict = {'000' : '0', '001' : '1', '010' : '0', '011' : '1', '100' : '1', '101' : '0', '110' : '1', '111' : '0'}
      ) -> None | list:
    '''
    
    Cellular Automata

    Inputs:
        N: int - number of values (cells)
        T: int - number of iterations of the simulation
        rules: dict - simulation rules
        arr: list - values (cells)

    '''

    assert N or arr is not None, "The user didn't enter any values"
    assert isinstance(T, int) and isinstance(rules, dict) and (isinstance(N, int) or isinstance(arr, list)), "Wrong type of argument passed"
    assert T > 0, "Wrong value for T" 

    if N is None:
        assert set(arr) == {0, 1}
    elif arr is None:
        assert N > 0
    else:
        assert N == len(arr)


    def transform(x: np.ndarray, arr_to_pass: np.ndarray, rls: dict) -> np.ndarray:
        val, idx = x
        # print(f'{val}, {idx}, {arr_to_pass[idx-1 : idx+2][:, 0]}\n\n\n')
        
        if idx not in {0, arr_to_pass.shape[0]-1}:
            vals = arr_to_pass[idx-1 : idx+2, 0].astype(str).tolist()
        elif idx == 0:
            vals = ['0'] + arr_to_pass[idx : idx+2, 0].astype(str).tolist()
        elif idx == arr_to_pass.shape[0] - 1:
            vals = arr_to_pass[idx-1 :, 0].astype(str).tolist() + ['0']
        
        return int(rls[''.join(vals)])
    

    np.random.seed(42)
    
    if arr is None:
        arr = np.concatenate([np.random.randint(0, 2, (N, 1)), np.arange(N).reshape(-1, 1)], axis=1)
    else:
        N = len(arr)
        arr = np.concatenate([np.array(arr).reshape(-1, 1), np.arange(N).reshape(-1, 1)], axis=1)

    arr_cp = arr.copy()
    
    for t in range(T):
        arr_cp[:, 0] = np.apply_along_axis(func1d=transform, arr=arr, axis=1, arr_to_pass=arr, rls=rules)
        arr = arr_cp.copy()


    return arr[:, 0]
        


print(CA(T=10, arr=[1, 0, 1, 1 ,1]))
````

# Problem 3 (30 pts). Sales

Table: ```Sales```


| Column Name | Type  |
|-------------|-------|
| sale_id     | int   |
| product_id  | int   |
| year        | int   |
| quantity    | int   |
| price       | int   |

(sale_id, year) is the primary key (combination of columns with unique values) of this table.
Each row records a sale of a product in a given year.
A product may have multiple sales entries in the same year.
Note that the per-unit price.

Write a solution to find prfits that occurred in the first year each product was sold.

For each product_id, identify the earliest year it appears in the Sales table.

Return profits for that product in that year.

Return a table with the following columns: product_id, first_year, quantity, and price.
Return the result in any order.
Plot the sales for each product and year.
 

Example 1:

Input: 
Sales table:

| sale_id | product_id | year | quantity | price |
|---------|------------|------|----------|-------|
| 1       | 100        | 2008 | 10       | 5000  |
| 2       | 100        | 2009 | 12       | 5000  |
| 7       | 200        | 2011 | 15       | 9000  |


Output: 

| product_id | first_year | profit |
|------------|------------|--------|
| 100        | 2008       | 50000  |
| 200        | 2011       | 135000 |


Name the function ```FYS```

````python
def FYS(Sales: pd.DataFrame) -> pd.DataFrame:
    '''
    First-year sales.
    Find the sales for each product in the first year

    Inputs:
        Sales: pd.DataFrame - DataFrame with the data on sales

    Outputs:
        df: pd.DataFrame - DataFrame with the data on profits

    '''

    assert isinstance(Sales, pd.DataFrame), "Wrong input format"
    # assert set(Sales.columns) == {'sale_id', 'product_id', 'year', 'quantity', 'price'}, "Wrong columns in DataFrame"

    min_year = Sales[['product_id', 'year']].groupby(by='product_id').agg('min').reset_index()
    df = min_year.merge(Sales[['product_id', 'year', 'quantity', 'price']], on=['product_id', 'year'])
    # print(df)
    df['profit'] = df['quantity'] * df['price']
    # print(df)
    df.rename(columns={'year' : 'first_year'}, inplace=True)
    # print(df)
    return df[['product_id', 'first_year', 'profit']]


Sales = pd.DataFrame(data=[[1, 100, 2008, 10, 5_000], [2, 100, 2009, 12, 5_000], [7, 200, 2011, 15, 9_000]],
                     columns=['sale_id', 'product_id', 'year', 'quantity', 'price']
                    )
print(Sales)

FYS(Sales)
````

# Bonus task (40 pts). Monte-Carlo strikes again

Create a python class ```MC```, which takes 3 values as arguments: ```N```- the number of points to generate ```r``` - the radius of the circle and ```prec```- the precision of the output (number of digits after comma to leave)\
The class should have 2 methods: 
1. ```init_sim```- creates N points on the neccesary domain, takes 1 non-trivial argument ```out``` if True, return the array of generated points, if False- doesn't
2. ```start_sim```- begins measuring whether each point belongs to the circle of radius ```r``` or not, estimates $\large \pi$ with precision ```prec```. Returns said estimation and saves to the class

In addition, the class should have 2 more magic methods besides ```__init__```: ```__repr__```- outputs the string containing the number of points 
(what exactly is in the string is up to you) and ```__bool__```, which is True if the value of $\large \pi$ estimated is equal to the real value of $\large \pi$ up to precision ```prec```

````python
class MC():

    def __init__(self, N: int, r: float, prec: int) -> None:
        self.N = N
        self.r = r
        self.prec = prec
        self.pts = None
        self.pi = None

    def __repr__(self) -> str:
        return f'Monte-Carlo with {self.N} points'

    def __bool__(self) -> bool:
        if not self.pi:
            return True if round(np.pi, self.prec) == self.start_sim() else False
        else:
            return True if round(np.pi, self.prec) == self.pi else False

    def init_sim(self, out: bool) -> None | np.ndarray:
        self.pts = np.random.uniform(-self.r, self.r, (self.N, 2))
        if out:
            return self.pts
        else:
            return None

    def start_sim(self) -> float | None:
        if self.pts is not None:
            self.inc = 0
            for pt in self.pts:
                if (pt[0]**2 + pt[1]**2)**(.5) <= self.r:
                    self.inc += 1
                    
            self.pi = round(4 * self.inc / self.N, self.prec)
            return self.pi
    
        else:
            return None

a = MC(1_000_000, 1, 4)
````
