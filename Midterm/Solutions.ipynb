# Task 1

```python
def func_1(arr: list[float])->list[float]: #define the function
    arr_out = [] #define a new array for output
    for element in arr: #iterate over the array
        arr_out.append(round((element+5)*10, 3)) #increment and round each element
    return arr_out #return the output array
```

# Task 2

```python
def func_2(arr: list[float], x: float)->list: #define the function
    arr_out = [] #define the output array
    for element in arr: #iterate over the array
        if element == x: #check whether the element is equal to x
            arr_out.append(0) #add 0 if element is equal it x
        else: #if element isn't equal to x
            arr_out.append((element+10)/5) #add the element
    return arr_out #return the array
```

# Task 3

```python
def func_3(arr_1: list[float], arr_2: list[float])->list[float]: #define the function
    out_sum = 0 #define the output sum
    for element_1, element_2 in zip(arr_1, arr_2): #iterate over both arrays simultaneously with zip
        if element_1 == element_2: #check that elements are equal
            out_sum += element_1 + element_2 #add both elements to the sum as per the task
    return out_sum #return the sum
```

# Task 4

```python
def func_4(arr_1: list[float], arr_2: list[float])->list[float|None]: #define the function
    #iterate over both arrays taking the index of the element
    for idx, (element_1, element_2) in enumerate(zip(arr_1, arr_2)):
        #Here we can get a division by zero error, so we define the try-except block
        try: #try to divive normally
            arr_1[idx] = element_1 / element_2
        except: #set the division result to None
            print("There is an error!")
            arr_1[idx] = None
    return arr_1 #return the result of the elementwise division
```

```python
def func_5()->list[str]: #define the function
    n = int(input("Please enter the number of iterations: ")) #get the number n from the task description
    arr_input = [] #define the array to store the user output
    for idx in range(n): #iterate n times to ask for input
        arr_input.append(input(f"Enter input number {idx}: ")) #add the user's input to your array
    return arr_input #return the array of user inputs
```

# Task 6

```python
def func_6()->None: #define the function
    with open("logs.txt", "a") as file: #open the file to write the output into
        
        var_to_enter = ["name", "sex", "age"] #define the variable names the user should enter
        idx = 0 #define the index of the loop
        inputs = [] #define the array of lines of user input
        
        while True: #define the infinite loop
            user_input = input(f"Enter your {var_to_enter[idx%3]}: ") #read the input from the user 
            if user_input == "STOP": #check the stopping condition
                break #break out of the infinite loop if the stop condition is met
            else: #if we don't stop
                inputs.append(user_input) #add the user input to the inputs array 
                idx += 1 #increment the index
            
        lines = [] #define the array of lines 
        for id_line in range(len(inputs) // 3): # iterate over the to-be lines in the file
            name, sex, age = inputs[id_line*3:(id_line+1)*3] #unpack the values 
            lines.append(f'{name},{sex},{age}\n') #refactor them into a line
            
        file.writelines(lines) #write the lines in the file
    return None #return None, this line is optional
```

# Task 8

```python
def func_8(prices:list[float])->int|float: #define the function
    profits = [] #define the arrays of profits
    num_days = len(prices) #define the number of days 
    for idx in range(num_days): #iterate over the days forward
        for idy in range(idx+1, num_days): #iterate over all of the future days w.r.t. day idx
            profits.append(prices[idy]-prices[idx]) #add the computed profit to the list of profits
    return max(profits) #return the max profit we can extract from the possible options
```

```python
def func_9(nums:list[int])->int: #define the function
    for num in range(len(nums)+1): #iterate over all integer numbers in the range [0, n]
        if num not in nums: #check that the number is/ isn't in the array
            return num #return the number if it isn't in the array
```
